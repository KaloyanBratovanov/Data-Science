```python
def extract_list_column(dataframe, id_column, list_column):
    extracted_dataframe = dataframe[
        [id_column, list_column]
    ].explode(list_column)

    extracted_dataframe = extracted_dataframe.dropna(
        subset=[list_column]
    ).reset_index(drop=True)

    return extracted_dataframe
```

```python
anime_themes_dataframe = extract_list_column(
    anime_dataframe, "anime_id", "themes"
)

anime_demographics_dataframe = extract_list_column(
    anime_dataframe, "anime_id", "demographics"
)

manga_themes_dataframe = extract_list_column(
    manga_dataframe, "manga_id", "themes"
)

manga_demographics_dataframe = extract_list_column(
    manga_dataframe, "manga_id", "demographics"
)

print("Anime themes:", len(anime_themes_dataframe))
print("Anime demographics:", len(anime_demographics_dataframe))
print("Manga themes:", len(manga_themes_dataframe))
print("Manga demographics:", len(manga_demographics_dataframe))
```

```python
anime_type_statistics = anime_dataframe.groupby("type").agg(
    anime_count=("anime_id", "count"),
    average_score=("score", "mean"),
    median_score=("score", "median"),
    average_members=("members", "mean")
).round(2)

print(
    anime_type_statistics.sort_values(
        "average_score", ascending=False
    )
)
```

```python
anime_type_statistics = anime_dataframe.groupby("type").agg(
    anime_count=("anime_id", "count"),
    rated_count=("score", "count"),
    average_score=("score", "mean"),
    median_score=("score", "median")
).round(2)

anime_type_statistics["rated_percentage"] = (
    anime_type_statistics["rated_count"]
    / anime_type_statistics["anime_count"] * 100
).round(2)

print(
    anime_type_statistics.sort_values(
        "average_score", ascending=False
    )
)
```

```python
nime_grouped_statistics = anime_dataframe.groupby(
    ["type", "status"]
).agg(
    anime_count=("anime_id", "count"),
    rated_count=("score", "count"),
    average_score=("score", "mean")
).round(2)

print(
    anime_grouped_statistics.sort_values(
        "anime_count", ascending=False
    ).head(15)
)
```

### * Problem 13. Text processing
Try to normalize names, tags, etc. Think about whitespace, word forms, etc. You'll find yourself dealing with both English and Japanese, so be careful about the encoding (the columns should be Unicode-encoded) and special symbols / numbers / etc.

Perform any text processing that you might see fit.

```python
import unicodedata

def normalize_title(title):
    if pd.isna(title):
        return title

    title = unicodedata.normalize("NFKC", title)
    title = " ".join(title.split())
    title = title.casefold()

    return title

anime_dataframe["normalized_title"] = (
    anime_dataframe["title"].apply(normalize_title)
)

manga_dataframe["normalized_title"] = (
    manga_dataframe["title"].apply(normalize_title)
)

print(
    anime_dataframe[
        ["title", "normalized_title"]
    ].head(10).to_string(index=False)
)
```

```python
normalized_anime_titles = set(
    anime_dataframe["normalized_title"].dropna()
)

normalized_manga_titles = set(
    manga_dataframe["normalized_title"].dropna()
)

normalized_common_titles = (
    normalized_anime_titles & normalized_manga_titles
)

print("Common titles before normalization:", len(common_titles))
print("Common titles after normalization:", len(normalized_common_titles))

print(
    "New matches:",
    len(normalized_common_titles) - len(common_titles)
)
```

```python
def normalize_tags(tags):
    return [
        " ".join(unicodedata.normalize("NFKC", tag).split())
        for tag in tags
    ]

for column in ["genres", "themes", "demographics"]:
    anime_dataframe[column] = anime_dataframe[column].apply(normalize_tags)
    manga_dataframe[column] = manga_dataframe[column].apply(normalize_tags)

print("Anime genres:")
print(anime_dataframe["genres"].head(10).to_string(index=False))

print("\nManga genres:")
print(manga_dataframe["genres"].head(10).to_string(index=False))
```

```python
for column in ["genres", "themes", "demographics"]:
    all_tags = (
        anime_dataframe[column].explode().dropna().tolist()
        + manga_dataframe[column].explode().dropna().tolist()
    )

    unique_tags = set(all_tags)
    normalized_tags = {tag.casefold() for tag in unique_tags}

    anime_duplicates = anime_dataframe[column].apply(
        lambda tags: len(tags) != len(set(tags))
    ).sum()

    manga_duplicates = manga_dataframe[column].apply(
        lambda tags: len(tags) != len(set(tags))
    ).sum()

    print(f"\n{column}:")
    print("Unique tags:", len(unique_tags))
    print("Unique tags ignoring case:", len(normalized_tags))
    print("Anime records with duplicate tags:", anime_duplicates)
    print("Manga records with duplicate tags:", manga_duplicates)
```
